```python
import numpy as np
import matplotlib.pyplot as plt
plt.style.use('./deeplearning.mplstyle')
```

```python
x_train = np.array([1, 2])
y_train = np.array([300.0, 500.0]) 
print(f'The sqft in 1000 sqft {x_train}')
print(f'The cost {y_train}')
```

```python
print(f' {x_train.shape}')
m = x_train.shape[0]
print(f' The number of training examples are {m}')

# or

m = len(x_train)
print(f'The number of training examples {m}')
```

```python
i=0
x_i = x_train[i]
y_i = y_train[i]
print(f'x^{i}, y^{i} = {x_i}, {y_i}')
```

```python
plt.scatter(x_train, y_train, marker = 'x' , c ='r')
plt.title("Linear Regression")
plt.ylabel("House Prices in 1000s of Dollar")
plt.xlabel(" Size (100 Sqft)")
```

```python
w = 200
b = 100
def compute_model_output(x, w, b):
    """
    Computes the prediction of a linear model
    Args:
      x (ndarray (m,)): Data, m examples 
      w,b (scalar)    : model parameters  
    Returns
      f_wb (ndarray (m,)): model prediction
    """
    m = x.shape[0]
    f_wb = np.zeros(m)
    for i in range(m):
        f_wb[i] = w * x[i] + b
        
    return f_wb
```

```python
tmp_f_wb = compute_model_output(x_train, w, b,)

# Plot our model prediction
plt.plot(x_train, tmp_f_wb, c='b',label='Our Prediction')
plt.scatter(x_train, y_train, marker='x', c='r',label='Actual Values')

# Set the title
plt.title("Housing Prices")
# Set the y-axis label
plt.ylabel('Price (in 1000s of dollars)')
# Set the x-axis label
plt.xlabel('Size (1000 sqft)')
plt.legend()
plt.show()
```
